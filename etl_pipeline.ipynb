```python
# Step 1: Extract 
import pandas as pd
orders_df= pd.read_csv('olist_orders_dataset.csv')
orders_df.info()
```

```python
customers_df= pd.read_csv('olist_customers_dataset.csv')
```

```python
#Step 2:Transform
#Shows missing records per column
print(orders_df.isnull().sum())
```

```python
#Shows missing records per column
print(customers_df.isnull().sum())
```

```python
# Drop rows with any missing values
orders_df= orders_df.dropna()
```

```python
# Convert the purchase timestamp to datetime format
orders_df['order_purchase_timestamp'] = pd.to_datetime(orders_df['order_purchase_timestamp'])
```

```python
# Verify the new row count and data types
orders_df.info()
```

```python
# Create a new column extracting just the month and year
orders_df['order_month'] = orders_df['order_purchase_timestamp'].dt.strftime('%Y-%m')
```

```python
# Drop any duplicate rows across both datasets
orders_df=orders_df.drop_duplicates()
customers_df=customers_df.drop_duplicates()
```

```python
# Join the tables using the shared customer_id column
merged_df=pd.merge(orders_df,customers_df, on='customer_id', how='inner')
merged_df.head()
```

```python
#Step 3:Load
merged_df.to_parquet('cleaned_olist_data.parquet', index=False)
print('Data successfully saved as cleaned olist data.parquet')
```
